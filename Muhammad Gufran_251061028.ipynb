```
import numpy as np
import scipy.stats as stats

# 1. Menyiapkan Data (100 Sampel)
# Menggunakan random seed agar hasil konsisten, bisa diubah sesuai data Anda
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)  # Contoh: 100 data dengan mean=50, std=10

print("--- 1. DATA SAMPEL (100 Data Pertama) ---")
print(data[:99], "... (total 100 data)\n")

# a. Estimasi Titik dan Interval (Distribusi Z, 5% kesalahan / 95% Confidence)
n = len(data)
mean_point = np.mean(data)  # Estimasi titik (rata-rata sampel)
std_dev = np.std(data, ddof=1)  # Standar deviasi sampel

# Tingkat kesalahan (alpha) = 5% -> Tingkat kepercayaan = 95% (0.95)
alpha = 0.05
z_score = stats.norm.ppf(1 - alpha / 2)

# Hitung Margin of Error dan Interval Kepercayaan
margin_of_error = z_score * (std_dev / np.sqrt(n))
ci_lower = mean_point - margin_of_error
ci_upper = mean_point + margin_of_error

print("--- a. ESTIMASI TITIK DAN INTERVAL (Distribusi Z) ---")
print(f"Estimasi Titik (Rata-rata) : {mean_point:.4f}")
print(f"Nilai Z (alpha = 5%)        : {z_score:.4f}")
print(f"Margin of Error             : {margin_of_error:.4f}")
print(f"Interval Kepercayaan (95%)  : [{ci_lower:.4f}, {ci_upper:.4f}]\n")

# b1. Resampling dengan Metode Jackknife
jackknife_means = []
for i in range(n):
    # Buat sampel dengan membuang 1 data ke-i
    jack_sample = np.delete(data, i)
    jackknife_means.append(np.mean(jack_sample))

jackknife_mean_of_means = np.mean(jackknife_means)

print("--- b1. METODE JACKKNIFE ---")
print(f"Jumlah iterasi jackknife    : {n}")
print(f"Rata-rata dari hasil jackknife : {jackknife_mean_of_means:.4f}\n")

# b2. Resampling dengan Metode Bootstrapping (1000 Iterasi)
n_iterations = 1000
bootstrap_means = []

for _ in range(n_iterations):
    # Pengambilan sampel secara acak dengan pengembalian (bootstrap sample)
    boot_sample = np.random.choice(data, size=n, replace=True)
    bootstrap_means.append(np.mean(boot_sample))

bootstrap_mean_of_means = np.mean(bootstrap_means)

print("--- b2. METODE BOOTSTRAPPING ---")
print(f"Jumlah iterasi bootstrap    : {n_iterations}")
print(f"Rata-rata dari hasil bootstrap : {bootstrap_mean_of_means:.4f}")
```
